```python
import numpy as np
import torch
```

6.4.1 张量元素类型转换

```python
tensor1 = torch.tensor([1, 2, 3])
print(tensor1)
print(tensor1.dtype)
```

```python
tensor1 = tensor1.type(torch.float32)
print(tensor1)
print(tensor1.dtype)
```

```python
tensor1 = tensor1.double()
print(tensor1)
print(tensor1.dtype)
```

```python
tensor1 = tensor1.to(dtype=torch.complex64)
print(tensor1)
print(tensor1.dtype)
```

6.4.2 tensor和ndarray转换

```python
# tensor -> ndarray
tensor1 = torch.rand(3, 2)
print(tensor1)
```

```python
ndarray1 = tensor1.numpy()
print(ndarray1)
```

```python
tensor1[:, 0] = 10
print(tensor1)
```

```python
print(ndarray1)
```

```python
# 使用copy方法避免共享内存
ndarray2 = tensor1.numpy().copy()
print(ndarray2)
```

```python
tensor1[:, 0] = 5
print(tensor1)
print(ndarray1)
print(ndarray2)
```

```python
# 从ndarray转换为tensor
ndarray1 = np.random.randn(3, 2)
tensor1 = torch.from_numpy(ndarray1)
print(ndarray1)
print(tensor1)
```

```python
ndarray1[:, 0] = 5
print(ndarray1)
print(tensor1)
```

```python
tensor2 = torch.from_numpy(ndarray1.copy())
ndarray1[:, 0] = 7
print(tensor2)
print(tensor1)
```

```python
# 使用tengsor方法直接转换成张量tensor
ndarray1 = np.random.randn(3, 2)
tensor1 = torch.tensor(ndarray1)
print(tensor1)
print(ndarray1)
```

```python
ndarray1[:, 0] = 9
print(tensor1)
print(ndarray1)
```

```python
tensor1 = torch.tensor(10)
print(tensor1)
print(tensor1.item())
```

```python
tensor1 = torch.tensor([10])
print(tensor1)
print(tensor1.item())
```

```python
tensor1 = torch.tensor([[10]])
print(tensor1)
print(tensor1.item())
```
