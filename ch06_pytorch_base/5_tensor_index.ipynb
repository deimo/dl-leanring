```python
# 布尔索引
# 1. 选取符合条件的行：首行元素大于5
mask = tensor1[:,:, 0] > 5
```

```python
mask
```

```python
print(tensor1[mask])
print(tensor1[mask].size())
```

```python
# 选取符合条件的列：第二列元素大于5
mask = tensor1[:, 1, :] > 5
```

```python
mask
```

```python
tensor2 = tensor1.mT
```

```python
tensor2
```

```python
tensor2[mask]
```

```python
tensor2[mask].T
```

```python
# 选取符合条件的矩阵：<1, 2> > 5
mask = tensor1[:, 1, 2] > 5
```

```python
mask
```

```python
tensor1[mask]
```
