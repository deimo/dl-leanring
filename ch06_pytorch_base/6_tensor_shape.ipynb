```python
## 张量拼接
```

```python
# 堆叠
tensor1 = torch.randint(1, 10, (4, 3, 5))
tensor2 = torch.randint(1, 10, (4, 3, 5))
```

```python
tensor1
```

```python
tensor2
```

```python
tensor3 = torch.stack((tensor1, tensor2), dim=3)
```
